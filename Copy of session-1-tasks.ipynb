### Q2:- Write a program that will convert celsius value to fahrenheit.

### Q3:- Take 2 numbers as input from the user.Write a program to swap the numbers without using any special python syntax.

````python
# Write your code here
fnum = int(input("Enter first number: "))
snum = int(input("Enter the second number: "))
temp = fnum
fnum = snum
snum = temp
print(f"The swapped numbers are {fnum} and {snum}")
````

### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
# Write your code here
x1 = int(input("Enter x1: "))
x2 = int(input("Enter x2: "))
y1 = int(input("Enter y1: "))
y2 = int(input("Enter y2: "))
result = (((x2-x1)**2)+((y2-y1)**2))**0.5
print(f"The euclidian distance between point ({x1},{x2}) and ({y1}, {y2} is: {result})")
````

### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2





````python
# Write your code here
heads = int(input("Enter the number of heads: "))
legs = int(input("Enter the number of legs: "))
dog_legs = (legs-(heads*2))
dogs=dog_legs/2
hens = heads - dogs
print(f"There are {dogs} dogs and {hens} chickens")
````

### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

````python
# Write your code here]
n = int(input("Enter the sum of nth natural numbers you want to find: "))
sum = int((n*(n+1)*((2*n)+1))/6)
print(f"The sum of first {n} natural numbers is : {sum} ")
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
# Write your code here
a = int(input("enter first number : "))
b = int(input("enter second number : "))
d = b-a
n = int(input("enter the number of nth term to find : "))
result = (a+((n-1)*d))
print(f"The nth term of this arithmetic series is {result}")
````

### Q9:- Given 2 fractions, find the sum of those 2 fractions.Take the numerator and denominator values of the fractions from the user.

````python
# Write your code here
n1 = int(input("Enter numerator of first number : "))
d1 = int(input("Enter denominator of first number : "))
n2 = int(input("Enter numerator if second number : "))
d2 = int(input("Enter denominator of second number : "))
sum = (((n1*d2)+(n2*d1))/(d1*d2))
print(f"the sum of two fractions is : {sum}")
````

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
# Write your code here
ht = int(input("Enter the height of the milk tank : "))
wt = int(input("Enter the width of the milk tank : "))
lt = int(input("Enter the length of the milk tank : "))
vt = ht*lt*wt
hg = int(input("Enter the height of the glass : "))
rg = int(input("Enter the radius of the glass : "))
vg = 3.14*(rg**2)*hg
num_glasses = int(vt/vg)
print(f"{num_glasses} glasses can be obtained from the tank.")
````
